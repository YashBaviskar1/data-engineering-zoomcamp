```python
from pyspark.sql.functions import to_date, col

df.filter(
    to_date(col("tpep_pickup_datetime")) == "2025-11-15"
).count()
```

```python
from pyspark.sql.functions import col

df.filter(
    (col("tpep_pickup_datetime") >= "2025-11-15") &
    (col("tpep_pickup_datetime") < "2025-11-16")
).count()
```

```python
df.createOrReplaceTempView("taxi")

spark.sql("""
SELECT COUNT(*)
FROM taxi
WHERE DATE(tpep_pickup_datetime) = '2025-11-15'
""").show()
```

```python
from pyspark.sql.functions import col, unix_timestamp, max

df.select(
    max(
        (unix_timestamp(col("tpep_dropoff_datetime")) -
         unix_timestamp(col("tpep_pickup_datetime"))) / 3600
    ).alias("longest_trip_hours")
).show()
```

```python
zone_df = spark.read.option("header", True).csv("taxi_zone_lookup.csv")

zone_df.show(5)
```

```python
df.createOrReplaceTempView("trips")
zone_df.createOrReplaceTempView("zones")
```

```python
from pyspark.sql.functions import count

result = (
    df.join(zone_df, df.PULocationID == zone_df.LocationID)
      .groupBy("Zone")
      .agg(count("*").alias("trip_count"))
      .orderBy("trip_count")
)

result.show(1)
```
